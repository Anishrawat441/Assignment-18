# Task 1: Manual One - Hot Encoding
## 1. Select 5 short sentences or number you decides for sentences from the dataset.
## 2. Build a vocabulary of unique words.
## 3. Create a one-hot encoded matrix manually using Python lists/dictionaries.
## 4. Display:
## . Vocabulary
## . One-hot vectors for each sentence

```python
import pandas as pd
```

```python
df=pd.read_csv("IMDB Dataset.csv")
```

```python
sentences = [
    "One of the other reviewers has mentioned that after watching just 1 Oz episode you'll be hooked",
    "I thought this was a wonderful way to spend time on a too hot summer weekend, sitting in the air conditioned theater and watching a light-hearted comedy.",
    "The filming technique is very unassuming- very old-time-BBC fashion and gives a comforting, and sometimes discomforting",
    "This show was an amazing, fresh & innovative idea in the 70's when it first aired",
    "If you like original gut wrenching laughter you will like this movie"
]
```

```python
## Build a vocabulary
vocab = set()

for sentence in sentences:
    for word in sentence.lower().split():
        vocab.add(word)

vocab = list(vocab)
print("Vocabulary:", vocab)
```

```python
## Create one hot encoding
# Create index mapping
word_index = {word: i for i, word in enumerate(vocab)}

# One-hot encoding
one_hot_encoded = []

for sentence in sentences:
    vector = [0] * len(vocab)
    for word in sentence.lower().split():
        index = word_index[word]
        vector[index] = 1
    one_hot_encoded.append(vector)
```

```python
## Display results
print("\nVocabulary:")
print(vocab)

print("\nOne-Hot Encoded Vectors:")
for i, vec in enumerate(one_hot_encoded):
    print(f"{sentences[i]} -> {vec}")
```

# Task 4: Understanding word Frequency
## 1. Calculate total frequency of each word.
## 2. Identify:
## . Top 10 most frequent words
## . Least frequent words
## 3. Explain how BoW captures frequncy information.

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
```

```python
sentences
```

```python
## Create Bag of Words
vectorizer = CountVectorizer()

bow = vectorizer.fit_transform(sentences)

# Convert to DataFrame
df = pd.DataFrame(
    bow.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print(df)
```

```python
## Calculate Total Frequency of Each Word
word_frequency = df.sum(axis=0)

print("Word Frequencies:")
print(word_frequency)
```

```python
## Top 10 Most Frequent Words
top_words = word_frequency.sort_values(ascending=False).head(10)

print("\nTop 10 Most Frequent Words:")
print(top_words)
```

```python
## Least Frequent Words
least_words = word_frequency.sort_values().head(10)

print("\nLeast Frequent Words:")
print(least_words)
```

## Explanation: How Bag of Words Captures Frequency Information

### Bag of Words (BoW) represents each document as a vector of word counts.

### Each column represents a unique word in the vocabulary.
### Each row represents one document (or sentence).
### The value in each cell is the number of times that word appears in the corresponding document.
### By summing each column, we obtain the total frequency of each word across the entire dataset.

# Task 8 : BoW vs TF-IDF Comparision
## 1. Compare word weights for common words.
## 2.Idetify:
## .Words  with high TF-IDF
## . Words with low TF-IDF
## 3. Explain why TF-IDF down -weights common words.

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd
```

```python
sentences
```

```python
##Create BoW Representation
bow_vectorizer = CountVectorizer()

bow_matrix = bow_vectorizer.fit_transform(sentences)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=bow_vectorizer.get_feature_names_out()
)

print("Bag of Words")
print(bow_df)
```

```python
## Create TF-IDF Representation
tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(sentences)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

print("\nTF-IDF")
print(tfidf_df.round(3))
```

```python
## Compare Word Weights
comparison = pd.DataFrame({
    "BoW Frequency": bow_df.sum(),
    "Average TF-IDF": tfidf_df.mean()
})

print("\nComparison:")
print(comparison.sort_values(by="Average TF-IDF", ascending=False))
```

```python
## High and Low TF-IDF Words
avg_tfidf = tfidf_df.mean()

print("\nWords with Highest TF-IDF:")
print(avg_tfidf.sort_values(ascending=False).head(5))

print("\nWords with Lowest TF-IDF:")
print(avg_tfidf.sort_values().head(5))
```

# Why TF-IDF Down-Weights Common Words
## Bag of Words (BoW) counts how many times each word appears but treats all words equally.
## TF-IDF (Term Frequency–Inverse Document Frequency) reduces the weight of words that occur in many documents because they carry less distinguishing information.
## Rare or unique words receive higher weights since they help differentiate one document from another.

# 1. Difference between One-Hot Encoding and Bag of Words (BoW)
## One-Hot Encoding	Bag of Words (BoW)
## Represents each word as a binary vector.	Represents each document as a vector of word counts.
## Indicates only whether a word exists in the vocabulary.	Stores how many times each word appears in a document.
## Does not capture word frequency.	Captures word frequency.
## Mainly used for encoding categorical words.	Mainly used for text representation in NLP tasks.

# 2. Why N-Grams Increase Dimensionality
## N-Grams create combinations of consecutive words instead of using only single words.
## Each unique unigram, bigram, or trigram becomes a separate feature.
## As the value of N increases, the number of possible word combinations grows rapidly.
## Therefore, the vocabulary size and feature vector dimensions increase, making the representation more sparse.

# 3. When to Prefer TF-IDF over BoW

## Use TF-IDF when:

## Common words (e.g., the, is, and) should have less importance.
## You want to identify words that are most relevant to each document.
## Performing tasks such as document classification, information retrieval, search engines, or document similarity.
## The dataset is large and contains many repeated words.

# 4. Limitations of Count-Based Vectorization
## Ignores the order of words (except when using N-Grams).
## Does not capture the meaning or context of words.
## Produces high-dimensional and sparse feature vectors.
## Treats synonyms (e.g., car and automobile) as completely different words.
## Gives equal importance to common and informative words (in BoW).
## Performs poorly on very large vocabularies due to increased memory and computation requirements.
